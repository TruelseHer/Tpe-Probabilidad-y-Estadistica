```python
# Importamos las librerias que vamos a usar.
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Cargar el archivo CSV con los datos

df = pd.read_csv('datos.csv') # Leemos el archivo .csv
df_1_1 = df[['1.1.1 Cantidad de personas que definen qué se cocina en el hogar','1.1.2 Género de quién/es definen qué se cocina']]
df_1_1.head()

df_1_2 = df[['1.2.1 Cantidad de personas que se encargan mayoritariamente de las compras','1.2.2 Género de quién/es se encargan de las compras']]
#df_1_2.head()

df_1_3 = df[['1.3.1 Cantidad de personas que se encargan de preparar las comidas en el hogar','1.3.2 Género de quién se encarga de preparar las comidas en el hogar']]

#df_1_3.head()
```

```python
df_1_1_ficant = df_1_1['1.1.1 Cantidad de personas que definen qué se cocina en el hogar'].value_counts()

df_1_1_ficant
```

```python
conteo = df_1_1['1.1.2 Género de quién/es definen qué se cocina'].value_counts()

# 2. Lo convertís a DataFrame
df_1_1_figenero = pd.DataFrame(conteo)
df_1_1_figenero
```

```python
df_1_1_ficant.plot.bar(rot=0)
```

```python
df_1_1_figenero = df_1_1_figenero.rename(columns={'count':'fi'})


# 1. Definir el orden de las categorías (revisá si en tus datos es 'Varón' o 'Varon' sin tilde)
categorias = ['Mujer', 'Mujer - Varón', 'Varón', 'Se desconoce']

# 2. Reordenar directamente el índice actual y rellenar con 0 si alguna categoría no tiene datos
df_1_1_figenero = df_1_1_figenero.reindex(categorias)

# 3. Graficar
df_1_1_figenero['fi'].plot.pie(autopct='%1.1f%%', ylabel='', title='Género de quién/es definen qué se cocina')
plt.show()

df_1_1_figenero


```
